```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Duffing oscillator differential equation
def duffing_oscillator(state, t, delta, alpha, beta, gamma, omega):
    x, x_dot = state
    dxdt = x_dot
    dx_dotdt = gamma * np.cos(omega * t) - delta * x_dot - alpha * x - beta * x**3
    return [dxdt, dx_dotdt]

# Simulate Duffing oscillator
def simulate_duffing(t, delta, alpha, beta, gamma, omega, initial_state):
    state = odeint(duffing_oscillator, initial_state, t, args=(delta, alpha, beta, gamma, omega))
    return state[:, 0]  # Return only the x (position)

# Set time range for simulation
t = np.linspace(0, 100, 10000)

# Initial conditions
initial_state = [0, 0]

# Parameters for non-chaotic case
delta_nc = 0.2
alpha_nc = -1.0
beta_nc = 1.0
gamma_nc = 0.3
omega_nc = 1.0

# Parameters for chaotic case
delta_c = 0.2
alpha_c = -1.0
beta_c = 1.0
gamma_c = 0.9
omega_c = 1.0

# Simulate non-chaotic case
x_nc = simulate_duffing(t, delta_nc, alpha_nc, beta_nc, gamma_nc, omega_nc, initial_state)

# Simulate chaotic case
x_c = simulate_duffing(t, delta_c, alpha_c, beta_c, gamma_c, omega_c, initial_state)

# Plot time series comparison
plt.figure(figsize=(14, 6))
plt.subplot(2, 2, 1)
plt.plot(t, x_nc, label='Non-chaotic')
plt.title('Non-chaotic Duffing Oscillator')
plt.xlabel('Time')
plt.ylabel('Position x')

plt.subplot(2, 2, 2)
plt.plot(t, x_c, label='Chaotic', color='r')
plt.title('Chaotic Duffing Oscillator')
plt.xlabel('Time')
plt.ylabel('Position x')

# Plot phase space comparison
plt.subplot(2, 2, 3)
plt.plot(x_nc[:-1], np.diff(x_nc) / np.diff(t), label='Non-chaotic')
plt.title('Phase Space (Non-chaotic)')
plt.xlabel('x')
plt.ylabel('dx/dt')

plt.subplot(2, 2, 4)
plt.plot(x_c[:-1], np.diff(x_c) / np.diff(t), label='Chaotic', color='r')
plt.title('Phase Space (Chaotic)')
plt.xlabel('x')
plt.ylabel('dx/dt')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# Lorenz system differential equations
def lorenz_system(state, t, sigma, rho, beta):
    x, y, z = state
    dxdt = sigma * (y - x)
    dydt = x * (rho - z) - y
    dzdt = x * y - beta * z
    return [dxdt, dydt, dzdt]

# Simulate Lorenz system
def simulate_lorenz(t, sigma, rho, beta, initial_state):
    state = odeint(lorenz_system, initial_state, t, args=(sigma, rho, beta))
    return state

# Set time range for simulation
t = np.linspace(0, 50, 10000)

# Initial conditions
initial_state = [1.0, 1.0, 1.0]

# Parameters for non-chaotic case
sigma_nc = 10.0
rho_nc = 22.0
beta_nc = 8/3

# Parameters for chaotic case
sigma_c = 10.0
rho_c = 28.0
beta_c = 8/3

# Simulate non-chaotic case
state_nc = simulate_lorenz(t, sigma_nc, rho_nc, beta_nc, initial_state)

# Simulate chaotic case
state_c = simulate_lorenz(t, sigma_c, rho_c, beta_c, initial_state)

# Plot time series comparison
plt.figure(figsize=(14, 8))

plt.subplot(2, 2, 1)
plt.plot(t, state_nc[:, 0], label='Non-chaotic')
plt.title('Non-chaotic Lorenz System (x component)')
plt.xlabel('Time')
plt.ylabel('x')

plt.subplot(2, 2, 2)
plt.plot(t, state_c[:, 0], label='Chaotic', color='r')
plt.title('Chaotic Lorenz System (x component)')
plt.xlabel('Time')
plt.ylabel('x')

# Plot phase space comparison
plt.subplot(2, 2, 3)
plt.plot(state_nc[:, 0], state_nc[:, 2], label='Non-chaotic')
plt.title('Phase Space (Non-chaotic)')
plt.xlabel('x')
plt.ylabel('z')

plt.subplot(2, 2, 4)
plt.plot(state_c[:, 0], state_c[:, 2], label='Chaotic', color='r')
plt.title('Phase Space (Chaotic)')
plt.xlabel('x')
plt.ylabel('z')

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# Generate periodic neural data: Rows are neurons, columns are time points
num_neurons = 10
num_timepoints = 200
time = np.linspace(0, 2 * np.pi, num_timepoints)  # 1 cycle over the time period

neural_data = np.zeros((num_neurons, num_timepoints))

# Create periodic signals for each neuron with different frequencies, phases, and amplitudes
for neuron in range(num_neurons):
    frequency = np.random.uniform(0.5, 2.0)  # Random frequency for each neuron
    phase = np.random.uniform(0, 2 * np.pi)  # Random phase for each neuron
    amplitude = np.random.uniform(0.5, 1.5)  # Random amplitude for each neuron
    neural_data[neuron, :] = amplitude * np.sin(frequency * time + phase)

# Performing PCA
pca = PCA(n_components=num_neurons)  # Keep all components
pca.fit(neural_data.T)  # Fit PCA on transposed data

# Get eigenvalues (variance explained by each principal component)
eigenvalues = pca.explained_variance_

# Plotting the eigenspectrum (sorted eigenvalues)
plt.figure()
plt.bar(range(1, len(eigenvalues) + 1), eigenvalues)
plt.xlabel('Principal Component')
plt.ylabel('Eigenvalue (Variance)')
plt.title('Eigenspectrum of Neural Data')
plt.show()
```
